```python
import numpy as np

# ==========================================
# INPUT NIM
# ==========================================
X = 5
Y = 6

print(f"Menggunakan nilai X = {X} dan Y = {Y}\n")
print("="*50)
print("BAGIAN 1: MATRIKS P dan Q (4x4)")
print("="*50)

# Inisialisasi Matriks P dan Q
P = np.array([
    [1, 2, X, -1],
    [3, 0, 1, 4],
    [-1, 2, 5, Y],
    [0, 1, -2, 3]
])

Q = np.array([
    [4, X, -1, 2],
    [1, 3, 0, 1],
    [2, -1, 2, Y],
    [0, 5, 1, -3]
])

# 1. P + Q
p1 = P + Q
print("1. P + Q :\n", p1, "\n")

# 2. 3P - 2Q
p2 = 3*P - 2*Q
print("2. 3P - 2Q :\n", p2, "\n")

# 3. P x Q
p3 = np.matmul(P, Q)
print("3. P x Q :\n", p3, "\n")

# 4. P x Q - Q x P
p4 = np.matmul(P, Q) - np.matmul(Q, P)
print("4. P x Q - Q x P :\n", p4, "\n")

# 5. (P + Q^T) x P
Q_T = Q.T
p5 = np.matmul(P + Q_T, P)
print("5. (P + Q^T) x P :\n", p5, "\n")


print("="*50)
print("BAGIAN 2: MATRIKS A dan B")
print("="*50)

# Inisialisasi Matriks A dan B
# Catatan: Pada soal bagian 2, x dan y huruf kecil
x = X
y = Y

A = np.array([
    [1, y, 3],
    [-2, 6, x]
])

B = np.array([
    [1, 2],
    [5, 7],
    [3, -x]
])

# 1. A - B^T
B_T = B.T
b1 = A - B_T
print("1. A - B^T :\n", b1, "\n")

# 2. AB
b2 = np.matmul(A, B)
print("2. AB :\n", b2, "\n")

# 3. BA
b3 = np.matmul(B, A)
print("3. BA :\n", b3, "\n")

# 4. A^T x B^T
A_T = A.T
b4 = np.matmul(A_T, B_T)
print("4. A^T x B^T :\n", b4, "\n")

# 5. (BA)^T
b5 = b3.T
print("5. (BA)^T :\n", b5, "\n")
```
